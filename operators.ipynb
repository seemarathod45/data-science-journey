```python
##Arithmetic Opr
a=15
b=20
print(a+b)
print(a-b)
print(a*b)
print(a/b)
print(a%b)
print(a//b)
print(a**b)
```

```python
##conditional opr
x=20
y=25
print(x==y)
print(x>y)
print(x<y)
print(x!=y)
print(x<=y)
print(x>=y)
```

```python
##logical opr
x=25
y=66
print(x==y and y==x)
print(x!=y and x==y)
print(not(x<y and y<x))
```
