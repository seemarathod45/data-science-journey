```python
#if
#Wap to check given number is greater than 50.
num= int (input("Enter a number"))
if num>50:
    print(f"{num} is greater than 50")
```

```python
#if else
#Wap to check given number is greater than 50 or not
num= int (input("Enter a number"))
if num >50:
    print(f"{num} is greater than 50")
else:
    print(f"{num} is not greater than 50" )
    
```

```python
#if elif else
#Wap to check given number is +ve, -ve or 0 
x=int (input("Enter a number"))
if x>0:
    print(f"{x} is +ve number ")
elif x<0:
    print(f"{x} is -ve number")
else :
    print(f"x is 0 number")
    
```

```python
#Nested If
#Wap to check age validation for voting
age=int(input("Enter your Age"))
if age>0 and age<100 :
    if age>18:
        print(f"your age is {age} Eligible for voting")
    else :
        print(f"your age is {age} not Eligible for Voting")
else:
    print("Invalid Age plz enter age between 0 to 99")
```
